```python
import os
import pandas as pd
import file_tool as ft
import numpy as np


path = r'D:\20240409\fstd\forrsd'
wl_path = r'D:\20240412\wl.csv'
path_list = ft.find_file(path,'.csv')
filename = [os.path.basename(path_list[i])[:-4] for i in range(len(path_list))]####保留文件名
```

```python
def RSD_calculate(spec):#算RSD
    X_ave = np.mean(spec,axis=1)
    std = np.std(spec,axis=1)
    
    return std/X_ave*100
```

```python
all_RSD = []
for i in range(5,90,5):
    RSD = []
    for paths in path_list:
        raw_feat = pd.read_csv(paths,header=None).values.astype(np.float32)
        raw_wl = pd.read_csv(wl_path,header=None).values.astype(np.float32)
        select_wl = np.where(np.isin(raw_wl,852.075))[0]
        RSD.append(RSD_calculate(raw_feat[select_wl,1:i]))

    RSD = np.array(RSD)
    all_RSD.append(RSD)
```

```python
all_RSD = np.array(all_RSD).reshape(17,-1)
all_data = np.concatenate((np.array(filename).reshape(-1,1).T,all_RSD),axis=0)
np.savetxt(r'D:\20240412\RSD.csv',all_data,delimiter=',',fmt='%s')
```
